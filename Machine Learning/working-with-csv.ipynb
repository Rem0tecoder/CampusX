Opening a csv file from an URL


```python
import requests 
from io import StringIO

url = ""
headers = {"User-Agent":""}
req = requests.get(url,headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

```python
def rename(name):
    if name == "Royal Challengers Banglore":
        return "RCB"
    else:
        return name
```
