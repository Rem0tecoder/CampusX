```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Creating Timestamp Objects

```python
# creating a timestamp
pd.Timestamp('2026/10/31')
```

```python
# variation
pd.Timestamp('2026-10-31')
```

```python
# only year
pd.Timestamp('2026')
```

```python
# using text
pd.Timestamp('31st October 2026')
```

```python
# Providing time
pd.Timestamp('31st October 2026 11:53AM')
```

```python
# using python datetime.datetime object
import datetime as dt

x = pd.Timestamp(dt.datetime(2026,10,31,11,55,44))
x
```

```python
# fetching attributes
x.year
x.month
x.day
x.hour
```

```python
date = np.array('2026-10-31', dtype=np.datetime64)
date
```

```python
date + np.arange(12)
```

Date-Time Index Object

A collection of pandas timestamp

```python
# from strings
pd.DatetimeIndex(['2026/1/1','2025/1/1','2024/1/1'])
```

```python
# using python datetime object
pd.DatetimeIndex(
    [dt.datetime(2026,1,1),
     dt.datetime(2025,1,1),
     dt.datetime(2024,1,1)])
```

```python
# using python timestamp object
dt_index = pd.DatetimeIndex(
    [pd.Timestamp(2026,1,1),
     pd.Timestamp(2025,1,1),
     pd.Timestamp(2024,1,1)])
```

```python
# using datetimeindex as series index
pd.Series([1,2,3],index=dt_index)
```

```python
# date_range function
pd.date_range(start='2023/1/5', end='2023/1/31',freq='2D')
```
